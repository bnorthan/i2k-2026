# Launching ND AI Lab


## What AI Lab does

- Image folders as a sequence
- Draw labels (napari tools, or SAM clicks)
- Label box: only boxed areas train
- Load previously trained models
- Segment, copy to labels, fix mistakes
- Augment
- train and predict

## Choose the project

| dataset | what it is |
|---|---|
| `bees` | bee images some labels |
| `ladybugs` | ladybug images no labels |
| `pollen_count` | pollen, works well with parameter tuning |


````python
DATASET = 'bees'   # 'bees' | 'ladybugs' | 'pollen_count'

from pathlib import Path

project_path = Path.cwd() / "data" / DATASET
if not project_path.is_dir():
    project_path = Path.cwd() / "notebooks" / "data" / DATASET

print(project_path)
````

## Launch

Before launching check what profiles we have. 

Profiles are used to switch between different option sets.  For example 2d-instance-skop, will only populate the options relevant to 2d instance segmentation and choose the scikit-ops versions. 

````python
import napari
from napari_ai_lab.apps.nd_ai_lab_launcher import launch_nd_ai_lab

from napari_ai_lab.apps.profiles import list_profiles, get_profile

list_profiles()
````

````python
viewer = napari.Viewer()

ai_lab, sequence_viewer, model = launch_nd_ai_lab(
    viewer,
    str(project_path),
    viewer_type="sequence",
    axes_to_collapse="C",
    axis_types="NYXC",
    profile="2d-instance-skop",
)

print('ND AI Lab launched on', project_path.name)
````

## Why those arguments

| argument | why |
|---|---|
| `viewer_type="sequence"` | all three sets hold images of different shapes, so they cannot be one stacked array |
| `axes_to_collapse="C"` | colour is not an axis to annotate along; labels are 2D per image |
| `axis_types="NYXC"` | N images, each Y by X with a colour axis |

The same three values suit all three datasets. A project of equal-sized
images could use `viewer_type="stacked"` instead.

## Using it

The **Sequence Viewer** at the bottom moves between images. The **AI Lab**
dock on the right has Label, Augment and Segment.

Interactive segmentation is in the Label tab. If you want to use the SAM3D option you need to have installed the full (not lite) environment.
